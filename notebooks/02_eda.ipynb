# Notebook 02: Exploratory Data Analysis (EDA)
**Goal:** Analyze skill frequency distribution, role categories, and text token densities.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

data_path = os.path.join("..", "data", "job_descriptions.csv")
df = pd.read_csv(data_path)
print("Job Categories Breakdown:")
print(df['category'].value_counts())
```

### Required Skills Token Analysis

```python
all_skills = []
for s_str in df['required_skills'].dropna():
    tokens = [s.strip() for s in s_str.split(',')]
    all_skills.extend(tokens)

skills_series = pd.Series(all_skills)
print("Top Skills by Demand in Dataset:")
print(skills_series.value_counts().head(6))
```
