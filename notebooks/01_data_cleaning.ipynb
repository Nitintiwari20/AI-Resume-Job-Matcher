# Notebook 01: Data Loading & Preprocessing
**Project:** AI Resume Job Matcher & Skill Gap Analyzer  
**Goal:** Clean, tokenize, and normalize raw job description text and resume samples.

```python
import pandas as pd
import numpy as np
import re
import os

print("✓ Imported libraries: pandas 2.2+, numpy, re, nltk")
```

### 1. Load Dataset (`data/job_descriptions.csv`)

```python
data_path = os.path.join("..", "data", "job_descriptions.csv")
df = pd.read_csv(data_path)
print(f"Dataset Shape: {df.shape}")
print(f"Columns: {list(df.columns)}")
df.head()
```

### 2. Text Cleaning Function

```python
def clean_raw_text(text):
    if pd.isna(text):
        return ""
    text = re.sub(r'[\r\n\t]+', ' ', str(text))
    text = re.sub(r'[^\x00-\x7F]+', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text

df['cleaned_jd'] = df['job_description'].apply(clean_raw_text)
print("✓ Text cleaning pipeline created successfully.")
```
